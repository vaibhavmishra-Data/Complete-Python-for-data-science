** **kwargs + normal parameter**

```python
def student(name, **details):
   print("Name",name)
   print("Other Details :", details)
```

```python
student(
    "priya",
    age = 30,
    city = "Delhi",
    course = "data science"
)
```

** **Kwargs + defaultparameter**

```python
def employee(name, department = "IT",**details):
   print("Name",name)
   print("Department" , department)

   print("Others", details)
```

```python
employee(
    "Ramesh",
    Salary = 30000,
    city = "Delhi",
    exp = 1
)
```
