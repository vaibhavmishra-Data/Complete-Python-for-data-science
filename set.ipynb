**Set operationes**

```python
compaign_A  = { "vaibhav", "aman", "priya", "neha"}
compaign_B  = { "aman", "priya", "karan", " simran"}
```

union

```python
all_customers = compaign_A.union(compaign_B)
print(all_customers)
```

intersection

```python
common = compaign_A.intersection(compaign_B)
print(common)
```

diffrence

```python
only_A = compaign_A.difference(compaign_B)
print(only_A)
```
