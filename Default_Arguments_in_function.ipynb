**Keyword Argument**

```python
def student(name, age):
   print(name)
   print(age)

student("Rahul", 20)
```

```python
student("Rahul",20)
```

**Postional Arguments**

```python
def introduce(name, age):
  print(name, age)

introduce("Rahul",20)
```

```python
def student(name, age):
   print("Name", name)
   print("Age", age)
```
