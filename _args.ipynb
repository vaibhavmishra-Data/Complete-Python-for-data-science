```python
def add(*args):
  total = 0
  for number in args:
    total = total + number
  return total
```

```python
print(add(10,20,30))
```

```python
print(add(10,20))
```

```python
def student(name, *marks):
  print("Name:", name)
  print("Marks:",marks)
```

```python
student("Priya",80,90,75)
```

```python
def student(name, *marks):
  total = sum(marks)
  average = total/len(marks)

  print("Name", name)
  print("Total", total)
  print("Average",average)
```

```python
student("Rahul",80,90,70,85)
```
